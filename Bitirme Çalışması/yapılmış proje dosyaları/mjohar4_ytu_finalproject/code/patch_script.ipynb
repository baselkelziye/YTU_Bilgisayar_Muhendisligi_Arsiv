```python
import pandas as pd
import numpy as np
import cv2 as cv
from PIL import Image,ImageDraw
# import openslide
import os
```

```python
#/content/drive/MyDrive/datasets/data
data_path = './images/'
mask_path = './masks/'
# check the patch dir if exist if not create ite

if not os.path.exists('./images_patches'):
    os.mkdir('./images_patches')

# check the mask patches dir if exist if not create ite

if not os.path.exists('./masks_patches'):
    os.mkdir('./masks_patches')

patches_path = './images_patches/'
masks_path = './masks_patches/'

# upload data
images_data = sorted(os.listdir(data_path),key=lambda x: int(x.split('.')[0]))

masks_data = sorted(os.listdir(mask_path),key=lambda x: int(x.split('.')[0]))

# Define the patch size and stride
patch_size = 512

```

```python
from patchify import patchify  # Import the patchify library
import tifffile as tif  # Import the tifffile library

# Iterate over each image in images_data
for image in images_data:
    img = tif.imread(data_path + image)  # Read the image using tifffile

    image_name = image.split(".")[0]  # Extract the image name without the extension

    patches_img = patchify(img, (512, 512, 3), step=512)  # Create patches of size 512x512 from the image
    # Save the patches as JPEG files
    for row in range(patches_img.shape[0]):
        for col in range(patches_img.shape[1]):
            single_patch = patches_img[row, col, :, :]  # Extract a single patch
            tif.imwrite(f'{patches_path}{image_name}_{row}_{col}.png', single_patch)  # Save the patch as a PNG file
            print(f'folder -> {patches_path}{image_name}_{row}_{col}.png --------> Done')
```
